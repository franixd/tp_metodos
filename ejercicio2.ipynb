```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
```

El par(α, β) determina la orientación de la familia y s selecciona uno de sus rayos

```python
#a)

def mascara_rayo(n, direccion, s):
    res =  np.zeros((n, n))
    for i in range(n) :
        for j in range(n):
            if (direccion[0]*i + direccion[1]*j) == s:
                res[i,j] =1
    return res
```

```python
#b) Visualizar las máscaras para las direcciones (1, 0), (0, 1), (1, 1), (1, −1) y (2, 1), 
# utilizando respectivamente los desplazamientos s = 3, 5, 7, 0, 8.

m1=mascara_rayo(8, (1,0), 3)
m2=mascara_rayo(8, (0, 1), 5)
m3=mascara_rayo(8, (1, 1), 7)
m4=mascara_rayo(8, (1, -1), 0)
m5=mascara_rayo(8, (2, 1), 8)

plt.imshow(m1)
plt.show()
plt.imshow(m2)
plt.show()
plt.imshow(m3)
plt.show()
plt.imshow(m4)
plt.show()
plt.imshow(m5)
plt.show()
```
