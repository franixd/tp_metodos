a)
La reconstrucción tomográfica es un problema inverso porque conocemos las mediciones b y la matriz A, que representa cómo los rayos atraviesan la imagen, y a partir de eso queremos recuperar la imagen desconocida x. El modelo b = Ax describe cómo una imagen genera las mediciones, pero en la reconstrucción hacemos el proceso inverso: usamos A y b para intentar encontrar x.

b) Para poder recuperar una única imagen, la matriz A tiene que tener rango columna completo, es decir, su rango debe ser igual a la cantidad de columnas. En ese caso, el espacio nulo de A tiene dimensión 0 y la única solución de Av = 0 es v = 0.

Si existe un vector no nulo v tal que Av = 0, entonces hay información de la imagen que A no puede detectar. En ese caso, x y x + v producirían las mismas mediciones, por lo que la reconstrucción no sería única.

c)
Cuando se agrega ruido, las mediciones b dejan de coincidir exactamente con alguna imagen, por lo que en general no existe un x que cumpla Ax = b de forma exacta. Por eso se usa mínimos cuadrados, que busca la imagen x que hace lo más chica posible la diferencia entre Ax y b.

El residuo representa esa diferencia entre las mediciones observadas y las que produce la reconstrucción. Sin embargo, una norma del residuo chica no garantiza que la imagen reconstruida sea correcta, porque distintas imágenes pueden producir mediciones muy parecidas o iguales, especialmente si existe espacio nulo.

d)
El término ||Dx||^2 supone que los píxeles vecinos de la imagen deberían tener valores parecidos, es decir, favorece imágenes más suaves.

Cuando lambda aumenta, la regularización tiene más peso y la reconstrucción queda más suave y con menos ruido. Sin embargo, si lambda es demasiado grande, también se pueden perder detalles importantes de la imagen. Con lambda chico se ajustan mejor las mediciones, pero puede quedar más ruido.

e)
np.linalg.lstsq obtiene directamente una solución de mínimos cuadrados sin tener que elegir una tasa de aprendizaje ni realizar iteraciones manuales. Gradiente descendiente, en cambio, necesita varias iteraciones y requiere elegir un valor de alfa adecuado para que converja. El método de Newton usa el gradiente y la Hessiana, y en este problema llega a la solución en una sola iteración porque la función de costo es cuadrática y la Hessiana es constante. En los tres casos, si están bien implementados, se obtiene prácticamente la misma solución.

f)
El modelo b = Ax relaciona la imagen x con las mediciones b a través de la matriz A. El rango permite analizar cuánta información independiente contienen las mediciones, mientras que el espacio nulo muestra qué cambios en la imagen no pueden ser detectados por A.

Cuando hay ruido, el sistema puede no tener una solución exacta y se usa mínimos cuadrados para encontrar la reconstrucción cuyas mediciones sean lo más cercanas posible a b. Geométricamente, esto corresponde a proyectar b sobre el espacio columna de A.

La regularización agrega información adicional al problema, favoreciendo imágenes más suaves mediante el término ||Dx||^2. Esto ayuda a reducir el efecto del ruido, aunque una regularización demasiado fuerte puede borrar detalles importantes.
