d) 
Una componente xk representa el coeficiente de atenuacion del pixel k. Los valores altos significan que la atenuacion del material es alta, por lo tanto al rayo le cuesta atravesar el material. Los valores bajos significan que la atenuacion del material es baja, es decir, el rayo pasa con facilidad al material. Y por ultimo, el valor nulo significa que el material no tiene atenuacion, entonces el rayo pasa intacto.
